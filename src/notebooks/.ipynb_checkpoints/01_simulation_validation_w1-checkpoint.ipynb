## 5. W1D2: Generate the Synthetic Raw Stream

`raw_stream.csv` represents the Machine/IoT Agent's input stream. The stream contains seeded synthetic laser-welding observations.

This notebook creates two groups:

1. **In-distribution synthetic cases:** scenarios produced by the project generator, including `normal`, `focal_offset`, gas, optical, vibration, speed, and power disturbances.
2. **OOD stress cases:** explicit `ood_focus_drift` cases generated by perturbing focus error, gas flow, vibration, contamination, and visual defects beyond the normal operating envelope.

The OOD cases are included because E1 requires normal/degradation/OOD separation.

```python
def attach_stream_metadata(
    df: pd.DataFrame,
    *,
    prefix: str,
    start_cycle: int,
    seed: int,
    quality_flag: str,
    provenance_source: str,
) -> pd.DataFrame:
    out = df.copy().reset_index(drop=True)
    out.insert(0, "record_id", [f"{prefix}_seed{seed}_{i:05d}" for i in range(len(out))])
    out.insert(1, "cycle", np.arange(start_cycle, start_cycle + len(out), dtype=int))
    start_time = pd.Timestamp("2026-09-03 00:00:00", tz="Asia/Shanghai") + pd.Timedelta(minutes=start_cycle)
    out.insert(2, "timestamp", pd.date_range(start_time, periods=len(out), freq="min").astype(str))
    out["provenance_source"] = provenance_source
    out["provenance_seed"] = seed
    out["quality_flag"] = quality_flag
    return out


def generate_ood_focus_drift(n_samples: int, seed: int, start_cycle: int) -> pd.DataFrame:
    rng = np.random.default_rng(seed + 2026)
    out = generate_laser_welding_data(n_samples=n_samples, seed=seed + 999)
    out["scenario"] = "ood_focus_drift"
    out["focal_position_error_mm"] += rng.uniform(0.55, 0.95, n_samples)
    out["shielding_gas_flow_l_min"] = np.clip(
        out["shielding_gas_flow_l_min"] - rng.uniform(6.0, 10.0, n_samples),
        3.0,
        None,
    )
    out["lens_contamination_level"] = np.clip(out["lens_contamination_level"] + rng.uniform(0.35, 0.65, n_samples), 0, 1)
    out["back_reflection_intensity"] = np.clip(out["back_reflection_intensity"] + rng.uniform(0.25, 0.45, n_samples), 0, 1)
    out["plume_intensity"] = np.clip(out["plume_intensity"] + rng.uniform(0.18, 0.35, n_samples), 0, 1)
    out["vibration_rms"] += rng.uniform(0.18, 0.35, n_samples)
    out["robot_path_error_mm"] += rng.uniform(0.12, 0.28, n_samples)
    out["porosity_risk"] = np.clip(out["porosity_risk"] + rng.uniform(0.25, 0.45, n_samples), 0, 1)
    out["visual_defect_score"] = np.clip(out["visual_defect_score"] + rng.uniform(25, 45, n_samples), 0, 100)
    out["time_since_lens_cleaning_h"] += rng.uniform(30, 60, n_samples)
    out = out.round(4)
    return attach_stream_metadata(
        out,
        prefix="w1_ood",
        start_cycle=start_cycle,
        seed=seed,
        quality_flag="synthetic_ood_unvalidated",
        provenance_source="synthetic_laser_welding_generator+ood_focus_drift_stress",
    )


base_stream = generate_laser_welding_data(n_samples=int(config["n_samples"]), seed=SEED)
base_stream = attach_stream_metadata(
    base_stream,
    prefix="w1_raw",
    start_cycle=0,
    seed=SEED,
    quality_flag="synthetic_unvalidated",
    provenance_source="synthetic_laser_welding_generator",
)
base_stream["synthetic_split"] = "in_distribution"

n_ood = max(30, int(0.10 * int(config["n_samples"])))
ood_stream = generate_ood_focus_drift(n_samples=n_ood, seed=SEED, start_cycle=len(base_stream))
ood_stream["synthetic_split"] = "ood"

raw_stream = pd.concat([base_stream, ood_stream], ignore_index=True)
raw_stream.to_csv(RAW_STREAM_PATH, index=False)
ood_stream.to_csv(OOD_STREAM_PATH, index=False)

print("Saved raw stream:", RAW_STREAM_PATH)
print("Saved OOD stream:", OOD_STREAM_PATH)
print("Raw stream shape:", raw_stream.shape)
raw_stream.head()
```

## 6. Raw Stream Validation Checks

These checks answer basic engineering questions:

- Are all expected feature columns present?
- Are there missing values in the feature contract?
- Is the stream reproducible with the same seed?
- Are metadata and provenance fields present?

Passing these checks does not prove physical realism. It proves the stream is usable and traceable enough for initial validation.

```python
base_stream_again = generate_laser_welding_data(n_samples=int(config["n_samples"]), seed=SEED)
base_stream_again = attach_stream_metadata(
    base_stream_again,
    prefix="w1_raw",
    start_cycle=0,
    seed=SEED,
    quality_flag="synthetic_unvalidated",
    provenance_source="synthetic_laser_welding_generator",
)
base_stream_again["synthetic_split"] = "in_distribution"

ood_stream_again = generate_ood_focus_drift(n_samples=n_ood, seed=SEED, start_cycle=len(base_stream_again))
ood_stream_again["synthetic_split"] = "ood"
raw_stream_again = pd.concat([base_stream_again, ood_stream_again], ignore_index=True)

required_metadata = [
    "record_id",
    "cycle",
    "timestamp",
    "scenario",
    "provenance_source",
    "provenance_seed",
    "quality_flag",
    "synthetic_split",
]

raw_checks = pd.DataFrame(
    [
        {"check": "all configured feature columns exist", "pass": all(col in raw_stream.columns for col in FEATURE_COLUMNS)},
        {"check": "all required metadata columns exist", "pass": all(col in raw_stream.columns for col in required_metadata)},
        {"check": "feature columns have no missing values", "pass": bool(raw_stream[FEATURE_COLUMNS].isna().sum().sum() == 0)},
        {"check": "record_id values are unique", "pass": bool(raw_stream["record_id"].is_unique)},
        {"check": "same seed reproduces the stream", "pass": bool(raw_stream_again.equals(raw_stream))},
        {"check": "OOD cases are present", "pass": bool((raw_stream["synthetic_split"] == "ood").any())},
        {"check": "normal cases are present", "pass": bool((raw_stream["scenario"] == "normal").any())},
        {"check": "focus-drift degradation cases are present", "pass": bool(raw_stream["scenario"].isin(["focal_offset", "ood_focus_drift"]).any())},
    ]
)
raw_checks.to_csv(RAW_CHECKS_PATH, index=False)

print("Saved raw checks:", RAW_CHECKS_PATH)
display(raw_checks)

raw_stream["scenario"].value_counts().rename_axis("scenario").reset_index(name="rows")
```

## 7. Scenario Plausibility Checks

For E1, the simulator must show sensible separation between normal, degradation, and OOD cases. The checks below compare scenario means against expected physical directions.

These are first-pass plausibility checks. They are not a substitute for laboratory calibration.

```python
def scenario_mean(scenario: str, column: str) -> float:
    values = raw_stream.loc[raw_stream["scenario"] == scenario, column]
    if values.empty:
        return float("nan")
    return float(values.mean())

normal_focal = scenario_mean("normal", "focal_position_error_mm")
focal_offset = scenario_mean("focal_offset", "focal_position_error_mm")
ood_focal = scenario_mean("ood_focus_drift", "focal_position_error_mm")
normal_gas = scenario_mean("normal", "shielding_gas_flow_l_min")
low_gas = scenario_mean("low_shielding_gas", "shielding_gas_flow_l_min")
ood_gas = scenario_mean("ood_focus_drift", "shielding_gas_flow_l_min")
normal_vibration = scenario_mean("normal", "vibration_rms")
fixture_vibration = scenario_mean("fixture_vibration", "vibration_rms")
normal_lens = scenario_mean("normal", "lens_contamination_level")
lens_proxy = scenario_mean("lens_contamination_proxy", "lens_contamination_level")
ood_lens = scenario_mean("ood_focus_drift", "lens_contamination_level")
normal_defect = scenario_mean("normal", "visual_defect_score")
ood_defect = scenario_mean("ood_focus_drift", "visual_defect_score")

plausibility_checks = pd.DataFrame(
    [
        {
            "expectation": "focal_offset has higher focal error than normal",
            "observed_comparison": f"{focal_offset:.3f} > {normal_focal:.3f}",
            "pass": bool(focal_offset > normal_focal),
        },
        {
            "expectation": "OOD focus drift exceeds in-distribution focal_offset",
            "observed_comparison": f"{ood_focal:.3f} > {focal_offset:.3f}",
            "pass": bool(ood_focal > focal_offset),
        },
        {
            "expectation": "low_shielding_gas has lower gas flow than normal",
            "observed_comparison": f"{low_gas:.3f} < {normal_gas:.3f}",
            "pass": bool(low_gas < normal_gas),
        },
        {
            "expectation": "OOD stress has lower gas flow than normal",
            "observed_comparison": f"{ood_gas:.3f} < {normal_gas:.3f}",
            "pass": bool(ood_gas < normal_gas),
        },
        {
            "expectation": "fixture_vibration has higher vibration than normal",
            "observed_comparison": f"{fixture_vibration:.3f} > {normal_vibration:.3f}",
            "pass": bool(fixture_vibration > normal_vibration),
        },
        {
            "expectation": "lens_contamination_proxy has higher lens contamination than normal",
            "observed_comparison": f"{lens_proxy:.3f} > {normal_lens:.3f}",
            "pass": bool(lens_proxy > normal_lens),
        },
        {
            "expectation": "OOD stress has higher lens contamination than normal",
            "observed_comparison": f"{ood_lens:.3f} > {normal_lens:.3f}",
            "pass": bool(ood_lens > normal_lens),
        },
        {
            "expectation": "OOD stress has higher visual defect score than normal",
            "observed_comparison": f"{ood_defect:.3f} > {normal_defect:.3f}",
            "pass": bool(ood_defect > normal_defect),
        },
    ]
)
plausibility_checks.to_csv(PLAUSIBILITY_PATH, index=False)

scenario_summary = raw_stream.groupby("scenario")[[
    "focal_position_error_mm",
    "shielding_gas_flow_l_min",
    "back_reflection_intensity",
    "plume_intensity",
    "vibration_rms",
    "robot_path_error_mm",
    "porosity_risk",
    "visual_defect_score",
    "lens_contamination_level",
]].mean().round(3)

print("Saved plausibility checks:", PLAUSIBILITY_PATH)
display(plausibility_checks)
display(scenario_summary)
```

## 8. W1D3: Build the Feature Vector

The Machine/IoT Agent role is represented here by the conversion from raw stream to model-ready feature vector.

In this Week 1 implementation, the feature vector contains:

- sensor/process features,
- metadata and provenance,
- synthetic split labels (`train`, `test`, `ood`),
- maintenance urgency target,
- scheduling context variables.

This prepares the evidence base for Week 2 prediction and uncertainty work.

```python
feature_vector = clean_laser_welding_data(raw_stream, FEATURE_COLUMNS)
feature_vector = add_maintenance_urgency(feature_vector)
feature_vector = add_scheduling_context(feature_vector, seed=SEED)

in_distribution = feature_vector[feature_vector["synthetic_split"] == "in_distribution"].copy()
train_df, test_df = train_test_split(
    in_distribution,
    train_fraction=float(config["train_fraction"]),
    seed=SEED,
)

train_ids = set(train_df["record_id"])
test_ids = set(test_df["record_id"])
feature_vector["split"] = "ood"
feature_vector.loc[feature_vector["record_id"].isin(train_ids), "split"] = "train"
feature_vector.loc[feature_vector["record_id"].isin(test_ids), "split"] = "test"
feature_vector["feature_vector_version"] = "w1_delivery_v1"
feature_vector["provenance_transform"] = "clean_laser_welding_data+add_maintenance_urgency+add_scheduling_context"
feature_vector["available_at"] = feature_vector["timestamp"]

feature_vector.to_csv(FEATURE_VECTOR_PATH, index=False)

print("Saved feature vector:", FEATURE_VECTOR_PATH)
print("Feature vector shape:", feature_vector.shape)
feature_vector.head()
```

## 9. Feature Vector Validation Checks

These checks verify that the feature vector can be used later by the AI Predictive Agent and scheduling environment.

```python
required_feature_metadata = [
    "record_id",
    "cycle",
    "timestamp",
    "available_at",
    "scenario",
    "split",
    "provenance_source",
    "provenance_seed",
    "quality_flag",
    "feature_vector_version",
    "provenance_transform",
]
required_targets = [
    "process_instability_score",
    "maintenance_urgency_score",
    "maintenance_label",
]
required_context = [
    "production_load",
    "resource_availability",
    "maintenance_cost_context",
]

feature_checks = pd.DataFrame(
    [
        {"check": "all configured features exist", "pass": all(col in feature_vector.columns for col in FEATURE_COLUMNS)},
        {"check": "all metadata fields exist", "pass": all(col in feature_vector.columns for col in required_feature_metadata)},
        {"check": "target fields exist", "pass": all(col in feature_vector.columns for col in required_targets)},
        {"check": "scheduling context fields exist", "pass": all(col in feature_vector.columns for col in required_context)},
        {"check": "feature columns have no missing values", "pass": bool(feature_vector[FEATURE_COLUMNS].isna().sum().sum() == 0)},
        {"check": "urgency score is within 0-100", "pass": bool(feature_vector["maintenance_urgency_score"].between(0, 100).all())},
        {"check": "train split is present", "pass": bool((feature_vector["split"] == "train").any())},
        {"check": "test split is present", "pass": bool((feature_vector["split"] == "test").any())},
        {"check": "ood split is present", "pass": bool((feature_vector["split"] == "ood").any())},
    ]
)
feature_checks.to_csv(FEATURE_CHECKS_PATH, index=False)

print("Saved feature checks:", FEATURE_CHECKS_PATH)
display(feature_checks)

feature_vector["split"].value_counts().rename_axis("split").reset_index(name="rows")
```

## 10. Target and Label Distribution

`maintenance_urgency_score` is a synthetic target derived from process instability. This is useful for controlled simulation, but it must be reported honestly.

Important limitation: a predictor trained in Week 2 may partly learn the urgency formula, not a real maintenance ground truth. The paper must describe this as simulation-based evidence.

```python
urgency_summary = feature_vector.groupby("split")["maintenance_urgency_score"].describe().round(3)
label_distribution = pd.crosstab(feature_vector["split"], feature_vector["maintenance_label"])
scenario_urgency = feature_vector.groupby("scenario")["maintenance_urgency_score"].agg(["count", "mean", "std", "min", "max"]).round(3)

display(urgency_summary)
display(label_distribution)
display(scenario_urgency.sort_values("mean", ascending=False))
```

## 12. Diagnostic Figure 2: Feature Distributions

This figure checks whether core features have usable distributions and whether OOD cases occupy a different region for stress-related variables.

## 13. W1D4: Scheduling Environment Sanity Check

The environment maps states to rewards through five possible maintenance actions.

State vector used by `LaserMaintenanceSchedulingEnv`:

| State index | Meaning |
| ---: | --- |
| 0 | urgency or ORCHESTRA-adjusted urgency |
| 1 | uncertainty |
| 2 | production load |
| 3 | resource availability |
| 4 | time since maintenance |
| 5 | maintenance cost context |
| 6 | process instability |

Action space:

| Action | Name |
| ---: | --- |
| 0 | do_nothing |
| 1 | inspect |
| 2 | minor_maintenance |
| 3 | major_maintenance |
| 4 | urgent_intervention |

```python
env_data = feature_vector[feature_vector["split"] != "ood"].reset_index(drop=True)
env = LaserMaintenanceSchedulingEnv(
    data=env_data,
    reward_weights=config["reward_weights"],
    episode_length=int(config["episode_length"]),
    seed=SEED,
)

initial_state = env.reset()
state_checks = {
    "state_shape_is_7": bool(initial_state.shape == (7,)),
    "state_values_are_finite": bool(np.isfinite(initial_state).all()),
    "state_values_within_expected_range": bool((initial_state >= 0).all() and (initial_state <= 1).all()),
}

low_row = env_data.sort_values("maintenance_urgency_score").iloc[0]
high_row = env_data.sort_values("maintenance_urgency_score").iloc[-1]
rows = []
for case_name, row in [("low_urgency_case", low_row), ("high_urgency_case", high_row)]:
    for action, action_name in ACTION_NAMES.items():
        result = env.evaluate_action(row, action, config["reward_weights"])
        rows.append(
            {
                "case": case_name,
                "case_urgency": float(row["maintenance_urgency_score"]),
                "action": action,
                "action_name": action_name,
                "reward": result.reward,
                "downtime": result.downtime,
                "cost": result.cost,
                "failure": result.failure,
                "unnecessary": result.unnecessary,
                "risk_after_action": result.risk_after_action,
                "alignment_bonus": result.alignment_bonus,
                "reward_is_finite": bool(np.isfinite(result.reward)),
            }
        )

environment_sanity = pd.DataFrame(rows)
environment_sanity.to_csv(ENV_SANITY_PATH, index=False)

print("Saved environment sanity table:", ENV_SANITY_PATH)
print(state_checks)
display(environment_sanity)
```

## 14. W1D5: Random and Rule-Based Baseline Smoke Test

This is not the final baseline comparison. It is a Week 1 smoke test to check that simple policies can run end-to-end with fixed seeds and finite outputs.

```python
random_rng = np.random.default_rng(SEED)


def random_policy(row: pd.Series) -> int:
    return int(random_rng.integers(0, 5))


baseline_smoke = pd.DataFrame(
    [
        evaluate_policy(
            "week1_random_smoke",
            env_data,
            random_policy,
            config["reward_weights"],
            n_episodes=5,
            episode_length=int(config["episode_length"]),
            seed=SEED,
        ),
        evaluate_policy(
            "week1_rule_based_smoke",
            env_data,
            rule_based_policy,
            config["reward_weights"],
            n_episodes=5,
            episode_length=int(config["episode_length"]),
            seed=SEED,
        ),
    ]
)
baseline_smoke.to_csv(BASELINE_SMOKE_PATH, index=False)

print("Saved baseline smoke table:", BASELINE_SMOKE_PATH)
baseline_smoke
```

## 15. W1D6: Rule-Based Trace and Reward Diagnostic

This trace is one full episode using the rule-based policy. Each row records the case, selected action, reward components, and operational context.

Week 3 will require stronger audit traces and counterfactuals. Week 1 only needs to show that a trace can be generated and inspected.

```python
env = LaserMaintenanceSchedulingEnv(
    data=env_data,
    reward_weights=config["reward_weights"],
    episode_length=int(config["episode_length"]),
    seed=SEED,
)

env.reset()
trace_rows = []
done = False
step_idx = 0
while not done:
    row = env._row()
    action = rule_based_policy(row)
    _, reward, done, info = env.step(action)
    trace_rows.append(
        {
            "step": step_idx,
            "record_id": row["record_id"],
            "cycle": int(row["cycle"]),
            "scenario": row["scenario"],
            "maintenance_urgency_score": float(row["maintenance_urgency_score"]),
            "process_instability_score": float(row["process_instability_score"]),
            "production_load": float(row["production_load"]),
            "resource_availability": float(row["resource_availability"]),
            "maintenance_cost_context": float(row["maintenance_cost_context"]),
            "action": int(action),
            "action_name": info["action_name"],
            "reward": float(reward),
            "downtime": float(info["downtime"]),
            "cost": float(info["cost"]),
            "failure": bool(info["failure"]),
            "unnecessary": bool(info["unnecessary"]),
            "risk_after_action": float(info["risk_after_action"]),
            "alignment_bonus": float(info["alignment_bonus"]),
        }
    )
    step_idx += 1

reward_trace = pd.DataFrame(trace_rows)
project_trace_path = PROJECT_LOGS_DIR / "week1_rule_based_trace.csv"
reward_trace.to_csv(project_trace_path, index=False)
reward_trace.to_csv(TRACE_EVIDENCE_PATH, index=False)

trace_checks = {
    "episode_length": len(reward_trace),
    "expected_episode_length": int(config["episode_length"]),
    "all_rewards_are_finite": bool(np.isfinite(reward_trace["reward"]).all()),
    "all_actions_are_valid": bool(reward_trace["action"].isin(ACTION_NAMES.keys()).all()),
}

print("Saved project trace:", project_trace_path)
print("Saved evidence trace:", TRACE_EVIDENCE_PATH)
print(trace_checks)
reward_trace.head()
```
